```python
# 5. Verify checkpoints landed on Drive
# (The training script unmounts Drive at the end, so we remount here)
import os

# Remount Drive if needed
if not os.path.exists('/content/drive/MyDrive'):
    from google.colab import drive
    drive.mount('/content/drive')

ckpt_dir = f"{PROJECT_PATH}/checkpoints/stage3_sentiment_large"

# Show Drive checkpoints
print('Drive checkpoints:')
if os.path.exists(ckpt_dir):
    files = [f for f in sorted(os.listdir(ckpt_dir)) if f.endswith('.pt')]
    if files:
        for f in files:
            size = os.path.getsize(os.path.join(ckpt_dir, f)) / 1e6
            print(f'  {f:35s} {size:8.1f} MB')
    else:
        print('  (no .pt files found)')
else:
    print(f'  Directory not found: {ckpt_dir}')
    print('  Training may not have reached the save stage.')

# Show local checkpoints (these survive Drive sync issues)
print('\nLocal checkpoints (/content/):')
local_files = [f for f in sorted(os.listdir('/content')) if f.endswith('.pt')]
if local_files:
    for f in local_files:
        size = os.path.getsize(f'/content/{f}') / 1e6
        print(f'  {f:35s} {size:8.1f} MB')
else:
    print('  (none - runtime may have been restarted)')

# Show latest log
import glob
local_logs = sorted(glob.glob('/content/stage3_sentiment_*.log'))
if local_logs:
    latest = local_logs[-1]
    print(f'\nLatest log (last 30 lines of {os.path.basename(latest)}):')
    with open(latest) as f:
        lines = f.readlines()
    for line in lines[-30:]:
        print(f'  {line.rstrip()}')
```
